# UBFC-rPPG Exploration

This notebook expects `MASKFUSIONNET_UBFC_ROOT` to point at a real UBFC-rPPG download. It has **not been run against real data** in this repository (no dataset is bundled) -- cell outputs below are empty on purpose. Run it yourself after setting up the dataset per `data/README.md`.

```python
import os, sys
sys.path.insert(0, '../src')
from maskfusionnet.data import discover_subjects, subject_level_split, UBFCrPPGDataset

root = os.environ['MASKFUSIONNET_UBFC_ROOT']
subjects = discover_subjects(root)
print(f'{len(subjects)} subjects found')
subjects
```

```python
split = subject_level_split(subjects, train_frac=0.7, val_frac=0.15, seed=42)
split.as_dict()
```

## Inspect one clip

```python
ds = UBFCrPPGDataset(root, split.train[:1], clip_length=160)
sample = ds[0]
sample['clip'].shape, sample['bvp'].shape, sample['hr_bpm']
```

```python
import matplotlib.pyplot as plt
plt.imshow(sample['clip'][:, 0].transpose(1, 2, 0))
plt.title('First frame of a clip (face-cropped)')
plt.axis('off')
```

```python
plt.plot(sample['bvp'])
plt.title(f"Ground-truth BVP, clip HR ~= {sample['hr_bpm']:.1f} bpm")
plt.xlabel('Frame')
```
